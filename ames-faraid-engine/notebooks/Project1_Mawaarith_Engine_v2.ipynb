## Charts

```python
fig,axes=plt.subplots(1,3,figsize=(17,5))
fig.patch.set_facecolor('#fdfaf3')
fig.suptitle('Mawārith Analytics v2 — PCIED Full Estate Statistics',fontsize=13,fontweight='bold',color='#0f3d1f')
awl=sum(1 for r in all_results if r.awl_applied)
radd=sum(1 for r in all_results if r.radd_applied)
tz=sum(1 for r in all_results if r.tanzeel_applied)
norm=20-awl-radd-tz
axes[0].pie([awl,radd,tz,norm],labels=["'Awl","Radd","Tanzeel","Standard"],
    autopct='%1.0f%%',colors=['#7b1fa2','#1565c0','#e65100','#1a6b2a'])
axes[0].set_title("Correction Type",fontweight='bold')
totals=[r.total_estate_value/1e6 for r in all_results]
axes[1].bar(range(1,21),totals,color='#1a6b2a',edgecolor='#c9961a',linewidth=0.8)
axes[1].set_xlabel('Case'); axes[1].set_ylabel('₦ Million'); axes[1].set_title('Total Estate per Case',fontweight='bold')
cash=[r.estate.cash/1e6 for r in all_results]
asst=[r.estate.asset_value/1e6 for r in all_results]
axes[2].bar(range(1,21),cash,label='Cash',color='#c9961a')
axes[2].bar(range(1,21),asst,bottom=cash,label='Assets',color='#1a6b2a')
axes[2].set_xlabel('Case'); axes[2].set_ylabel('₦ Million'); axes[2].set_title('Cash vs Assets',fontweight='bold')
axes[2].legend()
plt.tight_layout(); plt.show()
print(f"Total estates: ₦{sum(r.total_estate_value for r in all_results):,}")
print(f"Cash portion:  ₦{sum(r.estate.cash for r in all_results):,}")
print(f"Asset portion: ₦{sum(r.estate.asset_value for r in all_results):,}")
```
